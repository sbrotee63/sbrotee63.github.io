# Map of talk locations

This optional notebook runs `talkmap.py`, which reads published Markdown entries in `_talks/`. Install `PyYAML`, `geopy`, and `getorg` before running it. Entries without a location, unpublished examples, and duplicate locations are skipped. Geocoding uses a descriptive user agent and at least one second between requests. Generated map files go in `talkmap/`.

The sample map and page are excluded from the public site until genuine talk data is ready. To enable them, remove `published: false` from `_pages/talkmap.html`, remove `talkmap` from `_config.yml`'s exclude list, and set `talkmap_link: true` if the talks page should link to the map.


```python
from pathlib import Path
import runpy

root = Path.cwd()
if not (root / "talkmap.py").is_file():
    root = root.parent
if not (root / "talkmap.py").is_file():
    raise FileNotFoundError("Run this notebook from the repository root or _talks directory.")
runpy.run_path(str(root / "talkmap.py"), run_name="__main__")
```
