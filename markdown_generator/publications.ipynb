# Publications from TSV

This notebook runs `publications.py`, so the notebook and command-line generator use the same implementation. See `readme.md` for the input schema. Run from this directory or the repository root. Output files with matching names are overwritten. The bundled TSV examples stay unpublished.


## Data format


Required columns with non-empty values: `pub_date`, `title`, `venue`, `citation`, `url_slug`.

Optional columns: `excerpt`, `paper_url`, `slides_url`, `category`, `published`.

- Dates use `YYYY-MM-DD`.
- Each date and URL slug combination must be unique.
- Use the configured category keys: `books`, `manuscripts` (journal articles), or `conferences`. Blank categories become `uncategorized`, displayed under Other Publications.
- `published` accepts `true` or `false`; blank or omitted means true.
- Titles and other strings are escaped as YAML without changing quotes, Unicode, or HTML.
- Output uses `/publication/YYYY-MM-DD-url_slug`. The page layout displays citation and downloads; the excerpt is stored once.

## Reading and writing the data

The TSV uses tab separators because publication citations and talk descriptions often contain commas. You can edit the TSV with a spreadsheet or text editor. The standard-library CSV reader preserves empty cells, and JSON-compatible YAML quoting preserves Unicode, quotes, ampersands, and line breaks without adding HTML entities.

The generator reads every row, creates frontmatter and page content, then writes UTF-8 Markdown files in the sibling collection directory. Files with matching generated names are overwritten; duplicate date/slug combinations are rejected before writing. The notebook calls the same script as the command line to prevent the two versions from drifting.


```python
from pathlib import Path
import runpy
import sys

generator_dir = Path.cwd()
if not (generator_dir / "publications.py").is_file():
    generator_dir = generator_dir / "markdown_generator"
if not (generator_dir / "publications.py").is_file():
    raise FileNotFoundError("Run this notebook from the repository root or markdown_generator directory.")
if str(generator_dir) not in sys.path:
    sys.path.insert(0, str(generator_dir))
runpy.run_path(str(generator_dir / "publications.py"), run_name="__main__")
```
